```python
from litmol import PubChemClient
import pandas as pd
import json

data = pd.read_excel("corrected_material_glowing_data.xlsx")
data['glowing'] = data['glowing'].replace(2, 0)
data_cleaned = data.dropna(subset=['smiles'])

print(data_cleaned)

data_cleaned.to_csv('output.csv', index=False)
```

```python
smiles = data_cleaned['smiles']

with PubChemClient() as client:
    results = client.search_similar_batch(smiles)

with open("results.json", "w") as f:
    json.dump(results, f)
```

```python
rows = []
for input_smiles, similar_list in results.items():
    for sim in similar_list:
        rows.append({"input_smiles": input_smiles, "similar_smiles": sim})
flat_df = pd.DataFrame(rows)
flat_df.to_csv("similar_results.csv", index=False)
```
